Concatenation

```python
x1=pd.Series([10,200,300])
x2=pd.Series([200,5000,400,600],index=['a','b','c','d'])
print(pd.concat([x1,x2]))
```

```python
c=pd.Series([10,20,30])
new=pd.Series([15],index=[1])
s=pd.concat([c.iloc[:1],new,c.iloc[1:]]).reset_index(drop=True)
print(s)
```

```python
#create a series of 5 student marks and find average
#find highest and lowest
#print marks greater than 70
#create a series -->value ->marks and index->sub name
```

```python
import pandas as pd
# 1.
marks = pd.Series([85, 72, 65, 90, 78])
print("Marks:")
print(marks)
# 2.
print("Average:", marks.mean())
# 3.
print("Highest:", marks.max())
print("Lowest:", marks.min())
#or
print(marks.describe())
# 4.
print("Marks greater than 70:")
print(marks[marks > 70])
```

```python
#4.
subjectmarks = pd.Series([85, 72, 65, 90, 78],index=["Maths", "Science", "English", "Computer", "History"])
print("Subject wise Marks:")
print(subjectmarks)
```
